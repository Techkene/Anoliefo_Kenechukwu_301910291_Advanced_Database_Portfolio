## 9. Write cost: what the indexes cost on 100,000 inserts
Both inserts are rolled back, so the database is not changed. Compare the two `Time:` lines after each `INSERT 0 100000`.

```python
!psql -f write_cost_test.sql > results/write_cost_output.txt 2>&1
lines = open("results/write_cost_output.txt", encoding="utf-8").read().splitlines()
for i, l in enumerate(lines):
    if "===" in l:
        print(l.strip("= "))
    if l.startswith("INSERT"):
        print("   100,000 inserts took", lines[i + 1].replace("Time:", "").strip())
```

## 10. Numbers to copy into the README
Your timings will differ from mine because your computer is different. Update the results table in this folder's `README.md` with the numbers below, so the README matches your evidence.

```python
import psycopg2, pandas as pd
rows = [l for l in open("performance_results.md", encoding="utf-8").read().splitlines() if l.startswith("| Q") and not l.startswith("| Query")]
data = []
for l in rows:
    c = [x.strip() for x in l.strip("|").split("|")]
    data.append({"Query": c[0], "Before (ms)": c[3], "After (ms)": c[6], "Pages before": c[4], "Pages after": c[7], "Speed-up": c[9]})
pd.DataFrame(data)
```
